### W5A16 Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x


```python
def is_armstrong(x):
    s = str(x)
    n = len(s)
    tong = 0
    for chu_so in s:
        tong += int(chu_so) ** n
    return tong == x
n = int(input())
print(is_armstrong(n))
```

### W5A18 VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.


```python
def la_so_doi_xuung(n):
    if n < 0:
        return False
    so_ban_dau = n
    so_dao_nguoc = 0
    while n > 0:
        chu_so = n % 10
        so_dao_nguoc = so_dao_nguoc * 10 + chu_so
        n //= 10
    return so_ban_dau == so_dao_nguoc
print(la_so_doi_xuung(int(input())))
```

### W5A19 VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.


```python
def la_so_ky_dieu(n):
    if n <= 0:
        return False
    while n >= 10:
        tong_chu_so = 0
        while n > 0:
            tong_chu_so += n % 10
            n //= 10
        n = tong_chu_so
    return n == 1 
print(la_so_ky_dieu(int(input())))
```

### W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    try:
        if '.' in s:
            float(s)
            return 'float'
        else:
            int(s)
            return 'int'
    except ValueError:
        return 'neither'
print(detect_number_type(input()))
```

### W5A23 Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    delta = b**2 - 4 * a * c
    
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0
a, b, c = map(int, input().split())
print(quadratic_roots_count(a, b, c))
```

### W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.


```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
x1, y1, x2, y2, x3, y3 = map(int, input().split())
print(are_collinear(x1, y1, x2, y2, x3, y3))
```

### W5A27 Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def count_non_zero_digits(numbers):
    total_count = 0
    for n in numbers:
        n = abs(n)
        if n == 0:
            continue
        while n > 0:
            digit = n % 10
            if digit != 0:
                total_count += 1
            n //= 10
    return total_count
numbers = list(map(int, input().split()))
print(count_non_zero_digits(numbers))
```

### W5A29 Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min].

```python
def get_max_min(a, b, c):
    max_val = a
    if b > max_val:
        max_val = b
    if c > max_val:
        max_val = c
    min_val = a
    if b < min_val:
        min_val = b
    if c < min_val:
        min_val = c
    return [max_val, min_val]
a, b, c = map(int, input().split())
result = get_max_min(a, b, c)       
print(result[0], result[1])
```

### W5A30 Viết hàm tính hệ số nhị thức C(n, k).


```python
import math
def nCr(n, k):
    if k < 0 or k > n:
        return 0
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
n, k = map(int, input().split())
print(nCr(n, k))
```

### W5A34 Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau 

```python
def kiem_tra_chu_so_khac_nhau(x):
    s = str(x)
    return len(s) == len(set(s))

def sum_1_to_n(n):
    tong = 0
    for x in range(1, n + 1):
        if kiem_tra_chu_so_khac_nhau(x):
            tong += x
    return tong
n = int(input())
print(sum_1_to_n(n))    
```
